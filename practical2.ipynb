```python
import pandas as pd
data={
    "Name": ["Rahul","Vinit","Ruturaj","Himanshu"],
    "Age": ["38","20","27","19"],
    "Marks": ["50","80","40","60"],
    "Branch": ["CSE","AI","DS","AIML"]
    }
df=pd.DataFrame(data)
```

```python
print("Orignal dataframes",df)
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Marks']=df['Marks'].fillna(df['Marks'].median())
df['Branch']=df['Branch'].fillna('Unknown')
df
```

```python
import numpy as np
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')
df['Result'] = np.where(df['Marks'] > 66, 'Pass', 'Fail')
print(df)
```

```python
df.rename(columns={'Marks': 'score'}, inplace=True)
print("\nstatistical Summary:\n", df.describe())
```

```python
df['score'] = pd.to_numeric(df['score'], errors='coerce')
avg_score=df.groupby('Branch')['score'].mean()
print("average score by branch\n", avg_score)
```

```python
df['score'] = pd.to_numeric(df['score'], errors='coerce')
high_score=df[df['score']>80]
print("Student score above 80\n",high_score)
```

```python
print("Max score",np.max(df['score']))
```

```python
print("Min score",np.min(df['score']))
```

```python
print("Mean score",np.mean(df['score']))
```
