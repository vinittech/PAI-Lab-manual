# Practical 1: Object-Oriented Programming, Exception Handling & File Operations

This notebook covers three fundamental concepts in Python:
1. **Object-Oriented Programming (OOP)** - Classes, objects, inheritance, polymorphism
2. **Exception Handling** - Try-except blocks, custom exceptions
3. **File Operations** - Reading, writing, and manipulating files

```python
class Student:
  defdef __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def display(self):
    print(f"Name: {self.name}")
    print(f"Marks: {self.marks}")

  def result(self):
    return"Pass" if self.marks >= 40 else "Fail"

s1=Student("Rasna",85)
s2=Student("Amit",35)
s1.display()
print("Result:",s1.result())
s2.display()
print("Result:",s2.result())
```

```python
try:
  with open("student.txt","w") as file:
   file.write("Name: Rasna\n")
   file.write("Marks:85\n")
  print("Data written successfully")
except IOError:
  print("File write error")
```

```python
try:
  with open("student.txt","r") as file:
   content=file.read()
   print("File Content:\n",content)
except FileNotFoundError:
  print("file not found")
```
